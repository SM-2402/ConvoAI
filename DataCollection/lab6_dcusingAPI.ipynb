```python
import requests
import pandas as pd
```

```python
import os
from dotenv import load_dotenv

load_dotenv()

API_key = os.getenv("API_Key")
```

```python
#params are seperated by &

def get_data(API_key,year,sort_value,genre,pageno):
    url = 'https://api.themoviedb.org/3/discover/movie?api_key='+str(API_key)+'&primary_release_year.gte='+str(year)+'&sort_by='+str(sort_value)+'&with_genres='+str(genre)+'&page='+str(pageno)
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        return data['results']
    else:
        return 'error'
```

```python
get_data(API_key,2000,'popularity.desc','comedy',1) #each genre has an id,so just searching just by the name wont work
```

```python
get_data(API_key,2000,'popularity.desc',35,1)
```

```python
df = pd.DataFrame(columns=['Movie ID','Movie Name'])

pageno = 1
while(len(df)<300):
    query_results = get_data(API_key,2000,'popularity.desc',35,pageno)
    if query_results!= 'error':
        for movie in query_results: # it will pick each dictionary
            df.loc[len(df)] = [movie['id'],movie['title']] #len(df) :>initially zero, first movie :>1 ,.............

    pageno +=1
```

```python
def get_similar_movie(movie_id): #pageno is not added here as we only want 5 similar movies and the first page contains 20 movies
    url = 'https://api.themoviedb.org/3/movie/'+str(movie_id)+'/similar?api_key='+str(API_key)
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()
        return data['results']
    else:
        return 'error'
```

```python
df_similar = pd.DataFrame(columns=['Movie ID','Similar Movie ID'])

for movie_id in df['Movie ID']:
    query_response = get_similar_movie(movie_id)
    if query_response != 'error':
        for i in range(min(5,len(query_response))):
            df_similar.loc[len(df_similar)] = [movie_id,query_response[i]['id']]
```
