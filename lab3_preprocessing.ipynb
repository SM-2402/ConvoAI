Working with a custom dataset

```python
from nltk.corpus import PlaintextCorpusReader #A class in corpus library that is used to read plain text (.txt, .csv)
nltk.pathsec.ENFORCE = False
path = './imdb_text_dataset'

imdb = PlaintextCorpusReader(path,'.*') # '.*' : read any type of file it can read
```

```python
imdb.raw(fileids='review_1.txt')
```

```python
documents = []
filenames = imdb.fileids()

for filename in filenames: #adding the entire text of every file in the document list
    documents.append(imdb.raw(fileids=filename))
```

Preprocessing

S1: Tokenization and Normalization

```python
from nltk import word_tokenize #Better choice than str.split() method, because in diff languages the way words are split have different rules; splitting is needed to tokenize
```

```python
normalize_tokenize = []

for doc in documents: #word
    normalize_tokenize.append([word.lower() for word in nltk.word_tokenize(doc) if word.isalpha()]) # isalpha : checks whether if the words are alphabets or not, ||ly check isdigit,...
```

```python
normalize_tokenize[0]
```

M2: Stopword Removal

```python
stopword_list = stopwords.words('english')
nostop = []

for doc in normalize_tokenize:
    nostop.append([word for word in doc if word not in stopword_list])
```

```python
nostop
```

```python
len(normalize_tokenize[0]),len(nostop[0])
```

M4: Morphological Analysis

```python
from nltk.stem import PorterStemmer,LancasterStemmer
ps = PorterStemmer()
lS = LancasterStemmer()

preprocessed_text = []

for doc in nostop:
    preprocessed_text.append(''.join([ps.stem(word) for word in doc])) #.join is to convert the tokenized words into a single string
```

```python
preprocessed_text[0]
```

```python
from nltk.stem import WordNetLemmatizer
wn= WordNetLemmatizer()
preprocessed_text2 = []

for doc in nostop:
    preprocessed_text2.append(''.join([wn.lemmatize(word) for word in doc]))
```
