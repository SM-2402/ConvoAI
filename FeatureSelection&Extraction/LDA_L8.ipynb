# LDA

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

```python
dataset = load_iris()
df = pd.DataFrame(dataset.data,columns=dataset.feature_names)
df['target'] = dataset.target
```

```python
X = df.iloc[:,:-1]
Y= df.iloc[:,-1]
```

```python
def between_class(X,Y):
    n,k = X.shape
    global_mean = np.array(X.mean(axis=0)).reshape(1,k)
    classes = np.unique(Y)
    Sb = np.zeros((k,k),dtype=float)

    for cls in classes:
        X_subset = X[(Y==cls)]
        local_mean = np.array(X_subset.mean(axis=0)).reshape(1,k)
        diff = local_mean-global_mean
        Sb += X_subset.shape[0] * (diff.T@diff)
    return Sb
```

```python
Sb = between_class(X,Y)
Sb
```

```python
def within_class(X,Y):
    n,k = X.shape
    Sw = np.zeros((k,k),dtype=float)
    classes = np.unique(Y)

    for cls in classes:
        X_subset = X[(Y==cls)]
        local_mean = X_subset.mean(axis=0)
        diff = X_subset - local_mean
        Sw += diff.T@diff
    return np.array(Sw)
```

```python
Sw = within_class(X,Y)
Sw
```

```python
cov = np.linalg.inv(Sw)@Sb
values,vectors = np.linalg.eig(cov) # not eigh as Sw^-1Sb is not symm
```

```python
explained_variance = []

for value in values:
    explained_variance.append((value*100)/sum(values))
```

```python
explained_variance
```

```python
X_reduced = X@vectors[:,0]
```
