```python
import numpy as np
import pandas as pd
```

```python
R = df.pivot_table(index='user_id',columns='movie_id',values='rating',fill_value=0)
```

```python
R
```

```python
k = 50
values,vectors = np.linalg.eigh(R.T@R)
index = np.argsort(values)[::-1]
vectors = vectors[:,index]

V = vectors[:,:k]
S = np.maximum(np.sqrt(values[index]),0)

S = S[:k]

U = (R@V)/S
```

```python
R_low = U@np.diag(S)@V.T
```

```python
R_low
```

```python
movies = pd.read_csv(r"C:\Users\shash\Documents\college\ConvoAI\Data Pre-Processing-III-20261031\movies.csv")
movies
```

```python
def recommendation(user_id,topn):
    user_index = R.index.get_loc(user_id)
    scores = R_low.iloc[user_index]
    scores[(R.iloc[user_index]>0).to_numpy()] = -np.inf #(R.iloc[user_index]>0) will return a series -> convert it to a numpy array
    sort_indices = np.argsort(scores)[::-1]
    top_indices = sort_indices[:topn]
    recommendations = pd.DataFrame({'movie_id':R.columns[top_indices],
                                    'scores': scores[top_indices]})

    return recommendations.merge(movies,on='movie_id',how='left')
```

```python
recommendation(1,10)
```
